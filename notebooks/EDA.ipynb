```python
##step2: Categorical columns
```

```python
gender =df['gender'].value_counts()
gender
```

```python
import matplotlib.pyplot as plt
plt.xlabel('Gender')
plt.ylabel('Number of students')
plt.bar(gender.index,gender.values,width=0.5)
plt.show()
```

```python
race=df['race/ethnicity'].value_counts()
race
```

```python
plt.xlabel('Race/Ethnicity')
plt.ylabel('Number of students')
plt.bar(race.index,race.values)
plt.show()
```

```python
education=df['parental level of education'].value_counts()
education
```

```python
plt.xlabel('Parental Education level')
plt.ylabel('Number of students')
plt.bar(education.index,education.values)
plt.xticks(rotation=45)
plt.show()
```

```python
lunch=df['lunch'].value_counts()
lunch
```

```python
plt.xlabel('Lunch')
plt.ylabel('Number of students')
plt.bar(lunch.index,lunch.values)
plt.show()
```

```python
test_prep=df['test preparation course'].value_counts()
test_prep
```

```python
plt.xlabel('Test preparation')
plt.ylabel('Number of students')
plt.bar(test_prep.index,test_prep.values)
plt.xticks(rotation=45)
plt.show()
```
