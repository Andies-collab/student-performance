```python
##statistical summary
```

```python
math_min=df['math score'].min()
math_max=df['math score'].max()
math_mean=df['math score'].mean()
math_median=df['math score'].median()

print(math_min,math_max,math_mean,math_median)
```

```python
import matplotlib.pyplot as plt
plt.title("The Student's Math scores")
plt.xlabel('Maths scores')
plt.ylabel('Number of students')
plt.hist(df['math score'], bins=10,rwidth=0.95)
plt.show()
```

```python
read_min=df['reading score'].min()
read_max=df['reading score'].max()
read_mean=df['reading score'].mean()
read_median=df['reading score'].median()

print(read_min,read_max,read_mean,read_median)
```

```python
plt.title("The Student's Reading scores")
plt.xlabel('Reading scores')
plt.ylabel('Number of students')
plt.hist(df['reading score'], bins=10,rwidth=0.95)
plt.show()
```

```python
writing_min=df['writing score'].min()
writing_max=df['writing score'].max()
writing_mean=df['writing score'].mean()
writing_median=df['writing score'].median()

print(writing_min,writing_max,writing_mean,writing_median)
```

```python
plt.title("The Student's Writing scores")
plt.xlabel('Writing scores')
plt.ylabel('Number of students')
plt.hist(df['writing score'], bins=10,rwidth=0.95)
plt.show()
```
