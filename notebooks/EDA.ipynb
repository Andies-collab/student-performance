```python
education_order={
'some high school':0,
'high school':1,
'some college':2,
"associate's degree":3,
"bachelor's degree":4,
"master's degree":5

}
lunch_order={
    'free/reduced':0,
    'standard':1
}
test_prep={
    "none":0,
    "completed":1
}
df_encode= df
df['edu_encoded']=df_encode['parental level of education'].map(education_order)
df['lunch_encoded']=df_encode['lunch'].map(lunch_order)
df['testprep_encoded']=df_encode['test preparation course'].map(test_prep)



```

```python
dummies=pd.get_dummies(df['race/ethnicity'],dtype=int)
dummies
```

```python
dummy=pd.get_dummies(df['gender'],dtype=int)
```
